```python
from huggingface_hub import HfApi

username = "your_username_here"  # Replace with your username
repo_name = "your_repo_name"   # This will act as the root "folder"
repo_id = f"{username}/{repo_name}"

api = HfApi()

try:
    api.create_repo(repo_id=repo_id, private=True, repo_type="model")
    print(f"✅ Repo created: {repo_id}")
except Exception as e:
    print(f"⚠️ Repo may already exist: {e}")
```

### To save model folder

```python
from huggingface_hub import upload_file
import os

# Local model folder path
model_folder = "/content/drive/MyDrive/your_model_folder"
model_name_in_repo = "model_name_in_repo"  # This will be the subfolder name in HF

# Loop through each file in the folder
for file_name in os.listdir(model_folder):
    file_path = os.path.join(model_folder, file_name)

    if os.path.isfile(file_path):
        path_in_repo = f"{model_name_in_repo}/{file_name}"  # creates subfolder inside HF repo

        try:
            print(f"📤 Uploading: {path_in_repo}")
            upload_file(
                path_or_fileobj=file_path,
                path_in_repo=path_in_repo,
                repo_id=repo_id,
                repo_type="model"
            )
            print(f"✅ Uploaded: {path_in_repo}")
        except Exception as e:
            print(f"❌ Failed to upload {path_in_repo}: {e}")
```

### To save model file

```python
from huggingface_hub import upload_file
import os

# Local model file path
model_file = "/content/drive/MyDrive/your_model_file.pt"
model_name_in_repo = "model_name_in_repo"  # This will act as the subfolder name

# Extract just the file name
file_name = os.path.basename(model_file)
path_in_repo = f"{model_name_in_repo}/{file_name}"  # Simulates subfolder structure on HF

try:
    print(f"📤 Uploading: {path_in_repo}")
    upload_file(
        path_or_fileobj=model_file,
        path_in_repo=path_in_repo,
        repo_id=repo_id,
        repo_type="model"
    )
    print(f"✅ Uploaded: {path_in_repo}")
except Exception as e:
    print(f"❌ Failed to upload {path_in_repo}: {e}")
```
